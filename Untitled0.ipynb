```python
from google.colab import drive
drive.mount('/content/drive')

import zipfile
import os

zip_path = "/content/drive/MyDrive/DLOR ASSIGNMENT/DLOR_JELLYFISH_DATASET.zip"
extract_path = "/content/jellyfish_dataset"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted to:", extract_path)
```

```python
train_dir = "/content/jellyfish_dataset/Train_Test_Valid/Train"
valid_dir = "/content/jellyfish_dataset/Train_Test_Valid/valid"
test_dir  = "/content/jellyfish_dataset/Train_Test_Valid/test"
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = (64, 64)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True
)

valid_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

valid_gen = valid_datagen.flow_from_directory(
    valid_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

test_gen = test_datagen.flow_from_directory(
    test_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)
```

```python
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.layers import BatchNormalization, Dropout

def build_improved_model(input_shape=(64,64,3), num_classes=6):
    model = models.Sequential()

    # C1 - increased filters from 6 to 32
    model.add(layers.Conv2D(32, kernel_size=(5,5), activation='relu', padding='same', input_shape=input_shape))
    model.add(BatchNormalization())
    model.add(layers.MaxPooling2D(pool_size=(2,2)))

    # C3 - increased filters from 16 to 64
    model.add(layers.Conv2D(64, kernel_size=(5,5), activation='relu'))
    model.add(BatchNormalization())
    model.add(layers.MaxPooling2D(pool_size=(2,2)))

    # Flatten
    model.add(layers.Flatten())

    # F5 with Dropout to prevent overfitting
    model.add(Dropout(0.5))
    model.add(layers.Dense(120, activation='relu'))
    model.add(Dropout(0.5))
    model.add(layers.Dense(84, activation='relu'))

    # Output layer
    model.add(layers.Dense(num_classes, activation='softmax'))

    return model
```

```python
num_classes = train_gen.num_classes
model = build_improved_model(input_shape=(64,64,3), num_classes=num_classes)
model.summary()
```

```python
learning_rate = 0.001

model.compile(
    optimizer=optimizers.Adam(learning_rate=learning_rate),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

EPOCHS = 40*3

# Callbacks to prevent overfitting and improve training
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5)

history = model.fit(
    train_gen,
    validation_data=valid_gen,
    epochs=EPOCHS,
    callbacks=[early_stopping, reduce_lr]
)
```

```python
test_loss, test_acc = model.evaluate(test_gen)
print("TEST ACCURACY:", test_acc)
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

y_pred = model.predict(test_gen)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = test_gen.classes

print(classification_report(y_true, y_pred_classes, target_names=test_gen.class_indices.keys()))

cm = confusion_matrix(y_true, y_pred_classes)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```
